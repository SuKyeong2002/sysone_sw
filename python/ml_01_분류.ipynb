```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(15)
df = pd.DataFrame(np.random.randint(30, 100, 100), columns=['점수'])
df
```

```python
bins = [0, 60, 70, 80, 90, 101]
bin_label = list('가양미우수')
```

```python
df['등급'] = pd.cut(df['점수'], bins, right=False, labels=bin_label)
df
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

# 데이터 분할
x_df = df[['점수']] 
y_df = df['등급'] 

# 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(x_df, y_df, test_size=0.25, random_state=15)
```

```python
# 모델 학습 (만능 모델)
dt_clf = DecisionTreeClassifier(random_state=15)
dt_clf.fit(X_train.values, y_train.values.ravel())
```

```python
# 모델 예측
y_pred = dt_clf.predict(X_test.values)
```

```python
from sklearn.metrics import accuracy_score

# 정확도 평가
print(accuracy_score(y_test.values.ravel(), y_pred))
```

```python
# 모델 예측
label = dt_clf.predict([[75], [90], [81]])
print(label)
```
