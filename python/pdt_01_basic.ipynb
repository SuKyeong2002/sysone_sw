```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rc('font', family='gulim')
plt.rc('figure', figsize=(3, 3))
```

```python
x = [0, 1, 2, 3, 4]
y = [0, 4, 9, 16, 20]
x = np.arange(-10, 11, 1)
y = x**2
y1 = 3*x + 50
y2 = -x + 40
```

```python
plt.plot(x, y)
plt.plot(x, y1)
plt.plot(x, y2)
plt.show()
```

```python
plt.subplot(2, 2, 1)   # Axes 생성
plt.plot(x, y)
plt.title('subplot 1')

plt.subplot(2, 2, 2)
plt.plot(x, y1)
plt.title('subplot 2')

plt.subplot(2, 2, 3)
plt.plot(x, y2)
plt.title('subplot 3')

plt.subplot(2, 2, 4)
plt.plot(x, (lambda n:3*n**2)(x))
plt.title('subplot 4')

plt.tight_layout()
plt.show()
```

```python
plt.plot(x, y, 'o--b', x, y1, '+-r', x, y2, 's:k', linewidth=2, markersize=7)
plt.ylim(0, 22)
plt.title('제목')
plt.xlabel('x 축')
plt.ylabel('y 축')
plt.grid(True)
plt.legend(['1번', '2번', '3번'], loc='upper right')
plt.show()
```

```python
df = pd.DataFrame({'year':[n for n in range(2000,2010)],
                   'sales':np.random.randint(2000, 10000, size=10)})

df
```

```python
df.plot(x='year', y='sales', style='g--o', title="연도별 매출액")
plt.show()
```

```python
df.plot(x='year', y='sales', kind='bar', title="연도별 매출액")
plt.show()
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.barplot(x='day', y='tip', data=df)
plt.show()
```

```python
sns.barplot(x='day', y='tip', hue='sex', data=df)
plt.show()
```

```python
sns.lmplot(x='total_bill', y='tip', col='time', row='smoker', data=df)
plt.show()
```
